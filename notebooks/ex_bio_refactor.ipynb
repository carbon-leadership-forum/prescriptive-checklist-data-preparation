```python
test_df = pd.read_csv(test_file).reset_index(names="original_index")
```

```python
test_df_a1_a3 = test_df.iloc[::5].reset_index(drop=True)
test_df_a1_a3.index = 'element_'+test_df_a1_a3.index.astype(str)
test_df_a4 = test_df.iloc[1::5].reset_index(drop=True)
test_df_a4.index = 'element_'+test_df_a4.index.astype(str)
test_df_b2_b5 = test_df.iloc[2::5].reset_index(drop=True)
test_df_b2_b5.index = 'element_'+test_df_b2_b5.index.astype(str)
test_df_c2_c4 = test_df.iloc[3::5].reset_index(drop=True)
test_df_c2_c4.index = 'element_'+test_df_c2_c4.index.astype(str)
test_df_d = test_df.iloc[4::5].reset_index(drop=True)
test_df_d.index = 'element_'+test_df_d.index.astype(str)
```

```python
test_df_a1_a3_ex_bio = test_df_a1_a3.copy() 
test_df_b2_b5_ex_bio = test_df_b2_b5.copy() 
test_df_c2_c4_ex_bio = test_df_c2_c4.copy()
test_df_a1_a3_ex_bio_for_replacement_impacts = test_df_a1_a3.copy() 
test_df_a4_ex_bio_for_replacement_impacts = test_df_a4.copy() 
test_df_c2_c4_ex_bio_for_replacement_impacts = test_df_c2_c4.copy() 
```

```python
test_df_a1_a3_ex_bio["Global Warming Potential Total (kgCO2eq)"] = test_df_a1_a3_ex_bio[
    "Material Name"
    ].map(df_exBio_TallyGabi.set_index("name")["gwe_mfg"]) * test_df_a1_a3_ex_bio["Mass Total (kg)"]
test_df_a1_a3_ex_bio.loc[
    test_df_a1_a3_ex_bio["Material Name"].isin(incons_material_name),
    "Global Warming Potential Total (kgCO2eq)"
] = test_df_a1_a3["Global Warming Potential Total (kgCO2eq)"]

test_df_c2_c4_ex_bio["Global Warming Potential Total (kgCO2eq)"] = test_df_c2_c4_ex_bio[
    "Material Name"
    ].map(df_exBio_TallyGabi.set_index("name")["gwe_eol"]) * test_df_a1_a3_ex_bio["Mass Total (kg)"]
test_df_c2_c4_ex_bio.loc[
    test_df_c2_c4_ex_bio["Material Name"].isin(incons_material_name),
    "Global Warming Potential Total (kgCO2eq)"
] = test_df_c2_c4["Global Warming Potential Total (kgCO2eq)"]
```

```python
test_df_a1_a3_ex_bio_for_replacement_impacts["Global Warming Potential Total (kgCO2eq)"] = test_df_a1_a3_ex_bio_for_replacement_impacts[
    "Material Name"
    ].map(df_exBio_TallyGabi.set_index("name")["gwe_mfg"]) * test_df_b2_b5_ex_bio["Mass Total (kg)"]
test_df_a4_ex_bio_for_replacement_impacts["Global Warming Potential Total (kgCO2eq)"] = test_df_a4_ex_bio_for_replacement_impacts[
    "Global Warming Potential Total (kgCO2eq)"
    ] * test_df_b2_b5_ex_bio["Mass Total (kg)"] / test_df_a1_a3_ex_bio_for_replacement_impacts["Mass Total (kg)"]
test_df_c2_c4_ex_bio_for_replacement_impacts["Global Warming Potential Total (kgCO2eq)"] = test_df_c2_c4_ex_bio_for_replacement_impacts[
    "Material Name"
    ].map(df_exBio_TallyGabi.set_index("name")["gwe_eol"]) * test_df_b2_b5_ex_bio["Mass Total (kg)"]
test_df_b2_b5_ex_bio["Global Warming Potential Total (kgCO2eq)"] = (
    test_df_a1_a3_ex_bio_for_replacement_impacts["Global Warming Potential Total (kgCO2eq)"]
    + test_df_a4_ex_bio_for_replacement_impacts["Global Warming Potential Total (kgCO2eq)"] 
    + test_df_c2_c4_ex_bio_for_replacement_impacts["Global Warming Potential Total (kgCO2eq)"]
)
test_df_b2_b5_ex_bio.loc[
    test_df_b2_b5_ex_bio["Material Name"].isin(incons_material_name),
    "Global Warming Potential Total (kgCO2eq)"
] = test_df_b2_b5["Global Warming Potential Total (kgCO2eq)"]
```
